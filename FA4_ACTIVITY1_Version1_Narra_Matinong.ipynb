```python
# Asks for input
first_name = input("Enter first name: ")
last_name = input("Enter last name: ")
section = input("Enter section: ")
fav_subject = input("Enter favorite subject: ")

full_name = (first_name + last_name)
name_code = first_name.upper() + last_name.upper()
name_length = len(full_name)

# Prints output simply and withoutformatting
print("BASIC INFORMATION: ")
print("Name: ", first_name + last_name)
print("Name Code: ", name_code)
print("Section: ", section)
print("Favorite Subject: ", fav_subject)
print("Character Count: ", name_length)
```
