```python

# Asks for user's input
first_name = input("Enter first name: ")
last_name = input("Enter last name: ")
section = input("Enter section: ")
fav_subject = input("Enter favorite subject: ")

# Formats the input properly
full_name = first_name.strip().title() + " " + last_name.strip().title()
name_code = first_name.strip().upper() + " " + last_name.strip().upper()
upper_section = section.strip().upper()
favorite_subject = fav_subject.strip().title()
name_length = len(full_name.strip().replace(" ", ""))

# Prints the formatted results
print("FORMATTED INFORMATION: ")
print("Name: ", full_name)
print("Name Code: ", name_code)
print("Section: ", upper_section)
print("Favorite Subject: ", favorite_subject)
print("Character Count: ", name_length)
```
